## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset shows that 54.2% of pages are currently marked as declining. The median content age is 236 days, while the median time since the last update is 20 days. These numbers suggest there is enough variation in page performance and content age to make refresh prioritization worth exploring further. This is only an observed pattern in the starter data, not evidence that age or update timing directly causes decline.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import subprocess
import pandas as pd

REPO_DIR = "/content/flyrank-ml-internship"
REPO_URL = "https://github.com/HamzaKashif10/flyrank-ml-internship"

if not os.path.exists(REPO_DIR):
    subprocess.run(
        ["git", "clone", "--depth", "1", REPO_URL, REPO_DIR],
        check=True
    )

df = pd.read_csv(
    f"{REPO_DIR}/data/raw/content_refresh_anonymized.csv"
)

declining_rate = (df["trend_direction"].str.lower() == "down").mean()
median_age = df["content_age_days"].median()
median_update_gap = df["days_since_last_update"].median()

print(f"Declining pages: {declining_rate:.1%}")
print(f"Median content age: {median_age:.0f} days")
print(f"Median days since last update: {median_update_gap:.0f} days")
```
